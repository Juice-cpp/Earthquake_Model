```python
import xgboost as xgb
import numpy as np

print("Testando reconhecimento de GPU no XGBoost...")

try:
    # Criar dados fictícios muito pequenos
    X = np.random.rand(50, 2)
    y = np.random.randint(2, size=50)

    # Tentar treinar um modelo super simples forçando o uso da GPU
    clf = xgb.XGBClassifier(tree_method='hist', device='cuda', n_estimators=5)
    clf.fit(X, y)

    print("\n✅ SUCESSO! O XGBoost encontrou e usou a sua GPU sem problemas.")
except Exception as e:
    print("\n❌ ERRO! A GPU não foi reconhecida. Veja o erro abaixo:")
    print(e)
```
